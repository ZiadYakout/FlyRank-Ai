## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*


The starter dataset contains 30,000 pseudonymized content items across 32 clients. I observed that 54.2% of the content items have a downward trend, while 16,726 items have at least 500 impressions over the trailing 90 days. In addition, 13,152 items have both a downward trend and at least 100 impressions. These numbers suggest that there is a substantial set of pages with measurable search exposure and observable movement, making prioritization worth investigating. These observations do not prove that the pages need refreshing or that a refresh would cause recovery.

```python
down_pct = (df["trend_direction"] == "down").mean() * 100
high_visibility = (df["impressions_90d"] >= 500).sum()
down_with_demand = (
    (df["trend_direction"] == "down") &
    (df["impressions_90d"] >= 100)
).sum()

print(f"Pages in starter dataset: {len(df):,}")
print(f"Pages with downward trend: {down_pct:.1f}%")
print(f"Pages with >= 500 impressions: {high_visibility:,}")
print(f"Pages with >= 100 impressions and downward trend: {down_with_demand:,}")
```
